```python
import pandas as pd
import numpy as np

# -----------------------------------
# 1. Load Dataset
# -----------------------------------

df = pd.read_csv("sales_data.csv")

print("First 5 rows:")
print(df.head())

# Shape
print("\nDataset Shape:")
print(df.shape)

# Columns
print("\nColumns:")
print(df.columns)

# Data Types
print("\nData Types:")
print(df.dtypes)


# -----------------------------------
# 2. Data Cleaning
# -----------------------------------

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing numeric values with median
if df["Sales"].isnull().sum() > 0:
    df["Sales"] = df["Sales"].fillna(df["Sales"].median())

if df["Quantity"].isnull().sum() > 0:
    df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

# Fill missing category values with mode
if df["Category"].isnull().sum() > 0:
    df["Category"] = df["Category"].fillna(df["Category"].mode()[0])

# Remove duplicate rows
print("\nDuplicates before removal:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after removal:")
print(df.duplicated().sum())


# -----------------------------------
# 3. Basic Summary Statistics
# -----------------------------------

print("\nSales Summary Statistics:")

print("Mean:", df["Sales"].mean())
print("Median:", df["Sales"].median())
print("Minimum:", df["Sales"].min())
print("Maximum:", df["Sales"].max())


# -----------------------------------
# 4. NumPy Operation
# -----------------------------------

# Calculate average sales using NumPy
average_sales = np.mean(df["Sales"])

print("\nAverage Sales using NumPy:")
print(average_sales)


# -----------------------------------
# 5. Question 1
# Which category has the highest
# average sales?
# -----------------------------------

category_sales = df.groupby("Category")["Sales"].mean()

print("\nAverage Sales by Category:")
print(category_sales)

highest_category = category_sales.idxmax()

print("\nCategory with Highest Average Sales:")
print(highest_category)


# -----------------------------------
# 6. Question 2
# Which region has the highest
# total sales?
# -----------------------------------

region_sales = df.groupby("Region")["Sales"].sum()

print("\nTotal Sales by Region:")
print(region_sales)

highest_region = region_sales.idxmax()

print("\nRegion with Highest Total Sales:")
print(highest_region)


# -----------------------------------
# 7. Question 3
# Which products generated the
# highest sales?
# -----------------------------------

product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Product:")
print(product_sales)

print("\nTop 5 Products:")
print(product_sales.head(5))


# -----------------------------------
# 8. Additional Question
# Which products sold the highest
# quantity?
# -----------------------------------

product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop Products by Quantity Sold:")
print(product_quantity.head(5))


# -----------------------------------
# 9. Final Dataset Information
# -----------------------------------

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Dataset:")
print(df.head())
```
